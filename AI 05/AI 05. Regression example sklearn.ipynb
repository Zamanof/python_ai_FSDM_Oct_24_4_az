```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
x = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)
y = np.array([52, 57, 63, 70, 74])
```

```python
x
```

```python
model = LinearRegression()
model.fit(x, y)
```

```python
a = model.intercept_
b = model.coef_[0]
print(f"Tənlik y = {a:.2f} + {b:.2f} * x")
```

```python
y_pred = model.predict(x)
y_pred
```

```python
plt.scatter(x, y, color='red', label='Datalar')
plt.plot(x, y, color='blue', label='Reqressiya Datalar')
plt.xlabel('Hazırlanma saatı')
plt.ylabel('Qiymət')
plt.title('Sadə xətti reqressiya')
plt.legend()
plt.show()
```

```python
while True:
    try:
        x_new = float(input("Hazırlanma saatını daxil edin"))
        if x_new == 0:
            print("Çıxış")
            break
        y_new = a + b * x_new
        print(f"Proqnoz olunan qiymət = {y_new:.2f}")
        plt.scatter(x, y, color='red', label='Datalar')
        plt.plot(x, y, color='blue', label='Reqressiya Datalar')
        plt.scatter(x_new, y_new, color='red', label='Bizim proqnoz', zorder=5, s=100)
        plt.xlabel('Hazırlanma saatı')
        plt.ylabel('Qiymət')
        plt.title('Sadə xətti reqressiya')
        plt.legend()
        plt.show()
    except ValueError:
        print("Ədəd yazın")
```
