# Hesablama qrafı - Computational Graph
$$
y = x^2 + 5x + 7
$$

1. $x^2$ hesablanması
2. $5x$ hesablanması
3. Bu nəticələri cəmləyib üstünə 7 gəlirik




```python
X = torch.tensor(2.0, requires_grad=True)
Y = X ** 2 + 5 * X + 7
X, Y
```

```python
Y.grad_fn
```

```python
Y.backward()
print(X.grad)
```
