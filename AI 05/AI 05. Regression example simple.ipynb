```python
import numpy as np
import matplotlib.pyplot as plt
```

# Regression function

$$
   y = f(x_1, x_2, ..., x_n) + \epsilon
$$

- $f$ - asılılıq funksiyası
- $y$ - hədəf (target) dəyişən
- $x_i$ - xüsusiyyətlər (factor)
- $\epsilon$ - səhv(küy)

# Riyazi forma

$$
   y = a + bx + \epsilon
$$

- $y$ - hədəf (target) dəyişən
- $x$ - asılı olmayan(müstəqil) dəyişən (factor, feature)
- $a$ - sərbəst dəyişən (intercept)
- $b$ - əyilmə kooficienti (meyl əmsalı) (slope)
- $\epsilon$ - təasdüfi səhv(küy) - real nəticə və proqnozlaşdırılmış qiymət arasında fərq
    - $y$ - real nəticə Məslən: 70
    - $\hat{y}$ - proqnoz nəticəsində alınan qiymət Məsələn: 68.7
    - bu zaman $\epsilon=1.3$


```python
x = np.array([1, 2, 3, 4, 5])
y = np.array([52, 57, 63, 70, 74])
```

```python
n = len(x)
```

$$
y = a + bx
$$

$$
b = \frac{n \sum{xy} - \sum{x}\sum{y}}{n \sum{x^2} - (\sum{x})^2}
$$

$$
a = \frac{\sum{y} - b \sum{x}}{n}
$$

```python
b = (n*np.sum(x*y) - np.sum(x)*np.sum(y))/(n*np.sum(x**2) - np.sum(x)**2)
a = (np.sum(y) - b*np.sum(x))/n
```

```python
print(f"Tənlik y = {a:.2f} + {b:.2f} * x")
```
