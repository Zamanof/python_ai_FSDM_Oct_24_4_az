## Matplotlib `marker` dəyərləri

### Əsas marker-lər
- `.` — point
- `,` — pixel
- `o` — circle
- `v` — triangle down
- `^` — triangle up
- `<` — triangle left
- `>` — triangle right
- `1` — tri down
- `2` — tri up
- `3` — tri left
- `4` — tri right

### Kvadrat və çoxbucaqlılar
- `s` — square
- `p` — pentagon
- `h` — hexagon 1
- `H` — hexagon 2
- `8` — octagon

### Ulduz və romb
- `*` — star
- `D` — diamond
- `d` — thin diamond

### Xətt tipli marker-lər
- `+` — plus
- `x` — x
- `X` — filled x
- `P` — filled plus

### Digər
- `|` — vertical line
- `_` — horizontal line
- `None` — marker yoxdur
- `''` — marker yoxdur
- `' '` — marker yoxdur

```python
plt.scatter(x, y, color="red", label="Red", marker="<")
plt.scatter(y, x, color="blue", label="Blue", marker="4", s=sizes)
plt.show()
```
