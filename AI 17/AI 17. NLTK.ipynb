# 1. Tokenization

## 1.1 Word Tokenization (Sözlərə görə tokenizasiya)

````python
text = '''
Hello, world! My name is John O'Connor, and I'm a 25-year-old AI developer.

I live in New York, U.S.A. I work at OpenAI and earn $5,250.75 per month.

Today is 01/11/2026. My email is john.smith@gmail.com, and my website is https://example.com.

I love programming! Python, C++, and C# are my favorite languages.

Is AI changing the world? Absolutely!

Natural Language Processing (NLP) includes tokenization, stemming, and lemmatization.

Unbelievable! Tokenization can split words like "unhappiness" into smaller subwords.
'''
print(text.split())
````

````python
from nltk.tokenize import word_tokenize
tokens = word_tokenize(text)
print(tokens)
````

## 1.2 Sentence Tokenization (Cümlələrə görə tokenizasiya)

````python
from nltk.tokenize import sent_tokenize
sentences = sent_tokenize(text)

print('Sentences')

for i, sentence in enumerate(sentences, 1):
    print(f'{i}. {sentence.strip()}')
````

## 1.3 Digər növ tokenizasiyalar

````python
from nltk.tokenize import TreebankWordTokenizer, WordPunctTokenizer, RegexpTokenizer

text = "I can't believe it! John's 25-year-old car costs $5,000.50."

treebank_tokenizer = TreebankWordTokenizer()
print('TreebankWordTokenizer')
print(treebank_tokenizer.tokenize(text))

print()

word_punct_tokenizer = WordPunctTokenizer()
print('WordPunctTokenizer')
print(word_punct_tokenizer.tokenize(text))

print()

regex_tokenizer = RegexpTokenizer(r'\w+')
print('RegexpTokenizer')
print(regex_tokenizer.tokenize(text))
````

## 2. Lemmatization (Lemmatizasiya)

````python
from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

words = ['running', 'runs', 'ran', 'better', 'geese', 'mice', 'wolves', 'cats']
print('Nitq hissssələrini göstərmədən lemmatizasiya (by default - isim)')
print('-'*30)
for word in words:
    lemma = lemmatizer.lemmatize(word)
    print(f'{word:15} -> {lemma}')
````

````python
print('Nitq hissssələrini göstərərək lemmatizasiya')
# pos='v' - feil, pos='n' - isim, pos='a' - sifət, pos='r' - zərf, ...

print('-'*30)
print(f' running - verb -> {lemmatizer.lemmatize("running", pos="v")}')
print(f' runs - verb -> {lemmatizer.lemmatize("runs", pos="v")}')
print(f' ran - verb -> {lemmatizer.lemmatize("ran", pos="v")}')

print('-'*30)
print(f' better - adjective -> {lemmatizer.lemmatize("better", pos="a")}')

print('-'*30)
print(f' geese - noun -> {lemmatizer.lemmatize("geese", pos="n")}')
print(f' mice - noun -> {lemmatizer.lemmatize("mice", pos="n")}')
print(f' wolves - noun -> {lemmatizer.lemmatize("wolves", pos="n")}')
````

````python
print('Nitq hissssələrinin avtomatik olaraq təyin olunması ilə lemmatizasiya')
from nltk import pos_tag, FreqDist
from nltk.corpus import wordnet

def get_wordnet_pos(treebank_tag):
    if treebank_tag.startswith('J'):
        return wordnet.ADJ
    elif treebank_tag.startswith('V'):
        return wordnet.VERB
    elif treebank_tag.startswith('N'):
        return wordnet.NOUN
    elif treebank_tag.startswith('R'):
        return wordnet.ADV
    else:
        return wordnet.NOUN

text = """
John is a talented young programmer who lives in London.
He usually wakes up early and drinks hot coffee.
Every morning, he walks quickly to his office.
Yesterday, he developed an interesting application.
His colleagues were very happy with the results.
Today, John is working on a more difficult project.
He can solve complex problems, but he must work carefully.
Tomorrow, he will present his new project to the team.
Programming is exciting, and John loves his job.
"""

tokens = word_tokenize(text)
tagged = pos_tag(tokens)
print('Nitq hissələri')
print(tagged)

print('-' * 30)

for word, tag in tagged:
    wordnet_tag = get_wordnet_pos(tag)
    lemma = lemmatizer.lemmatize(word, get_wordnet_pos(tag))
    print(f'{word:15} ({tag:3}) {lemma}')
````

# 3. Stemming

````python
from nltk.stem import PorterStemmer, LancasterStemmer, SnowballStemmer

porter_stemmer = PorterStemmer()
lancaster_stemmer = LancasterStemmer()
snowball_stemmer = SnowballStemmer('english')

words = ['running', 'runs', 'easily', 'fairly', 'programming', 'programmer', 'developer', 'programs']

print(f'{"Söz":15} {"Porter":<12} {"Lancaster":<12} {"Snowball":<12}')

for word in words:
  p = porter_stemmer.stem(word)
  l = lancaster_stemmer.stem(word)
  s = snowball_stemmer.stem(word)
  print(f'{word:15} {p:<12} {l:<12} {s:<12}')
````

````python
# Stemming vs Lemmatization

print('-'*50)
print(f'{"Söz":15} {"Stemming":<15} {"Lemmatization":<15}')

comparison_words = ['studies', 'studying', 'better', 'running', 'wolves']

print('-'*50)
for word in comparison_words:
  p = porter_stemmer.stem(word)
  l = lemmatizer.lemmatize(word)
  print(f'{word:15} {p:<15} {l:<15}')
````

# 5. Mətnin ilkin emalının tam pipeline-ı

````python
def preprocess_text(text, use_lemmatization=True):
    text = text.lower()

    tokens = word_tokenize(text)

    stop_words = set(stopwords.words('english'))

    tokens = [token for token in tokens if token not in stop_words and token.isalpha()]

    if use_lemmatization:
        lem = WordNetLemmatizer()
        pos_tags = pos_tag(tokens)
        tokens = [lem.lemmatize(t, pos=get_wordnet_pos(pos)) for t, pos in pos_tags]
    else:
        stemmer = PorterStemmer()
        tokens = [stemmer.stem(token) for token in tokens]
    return tokens
````

````python
asample_text = """
John is a talented young programmer who lives in London.
He usually wakes up early and drinks hot coffee.
Every morning, he walks quickly to his office.
Yesterday, he developed an interesting application.
His colleagues were very happy with the results.
Today, John is working on a more difficult project.
He can solve complex problems, but he must work carefully.
Tomorrow, he will present his new project to the team.
Programming is exciting, and John loves his job.
"""

print('Original text:')
print(asample_text)

print('-' * 50)
print(preprocess_text(asample_text))

print('-' * 50)
print(preprocess_text(asample_text, False))
````
