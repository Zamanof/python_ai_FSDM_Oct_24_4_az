```python
import matplotlib.pyplot as plt
from fontTools.varLib.mutator import percents
```

```python
languages = ["It dili", "Adam dili", "İşarət dili", "Qarğa dili"]
words = [3, 100, 50, 200]
```

```python
plt.bar(languages, words, color="aqua")
plt.xlabel("Dillər")
plt.ylabel("Sözlər")
plt.title("Dillərin sözlərinin sayı")
plt.show()
```

```python
plt.barh(languages, words, color=["red", "blue", "orange", "#2ce6e6"])
plt.xlabel("Dillər")
plt.ylabel("Sözlər")
plt.title("Dillərin sözlərinin sayı")
for i, value in enumerate(words):
    plt.text(i, i+0.43, str(value), ha="left")
plt.show()
```
