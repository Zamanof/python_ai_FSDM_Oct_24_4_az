```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
x = np.arange(0, 10)
x
```

```python
y = x*2
y
```

```python
plt.plot(x, y)
plt.title("y = 2x")
plt.xlabel("x")
plt.ylabel("y")
plt.xlim(0, 10)
plt.ylim([0, 20])
plt.savefig("plot.png", dpi=300)
plt.show()
```

```python
fig = plt.figure()
axes = fig.add_axes([0, 0, 1, 1]) # left, bottom, width, height
axes.plot(x, y)
```

```python
a = np.linspace(0, 10, 15)
a
```

```python
b = a**2
b
```

```python
# fig = plt.figure(dpi=300, figsize=(2, 2))
fig = plt.figure()
axes1=fig.add_axes([0, 0, 1,1])
axes1.plot(a, b)
axes1.set_xlabel("x")
axes1.set_ylabel("y")
axes1.set_title("y = x^2")
axes1.set_xlim([0, 10])
axes1.set_ylim([0, 100])
axes2 = fig.add_axes([0.5, 0.5, 0.5, 0.5])
axes2.plot(x, y)
axes2.set_xlabel("x")
axes2.set_ylabel("y")
```

```python
fig, axes = plt.subplots(nrows=1, ncols=2)
axes[0].plot(a, b)
axes[1].plot(x, y)
plt.tight_layout()
```

```python
fig, axes = plt.subplots(nrows=2, ncols=2)
axes[0, 0].plot(a, b)
axes[0, 1].plot(a, np.sin(a))
axes[1, 0].plot(x, np.tan(x))
axes[1, 1].plot(x, y)
plt.tight_layout()
```

```python
fig = plt.figure()
axes = fig.add_axes([0, 0, 1, 1])
axes.plot(a, b, label="y=x^2", color="#e315eb")
axes.plot(a, np.sin(a), label="y=sin(x)", color="blue")
axes.plot(x, y, label="y=2x", color="green", lw=8)
axes.set_xlabel("x")
axes.set_ylabel("y")
axes.set_title("y = x^2, sin(x), 2x")
axes.legend(loc="upper right")
axes.set_xlim([0, 10])
axes.set_ylim([0, 100])

plt.show()
```
