# Pandas `describe()` statistik göstəriciləri

## `describe()`

`describe()` — `DataFrame` və ya `Series` daxilində ədədi məlumatların əsas statistik göstəricilərini göstərir.

```python
df.describe()
```

Nəticədə aşağıdakı göstəricilər alınır:

---

## `count`

**count** — boş olmayan dəyərlərin sayını göstərir.

`NaN` olan dəyərlər sayılmır.

Məsələn, sütunda 100 sətir varsa, amma 5 dəyər boşdursa:

```text
count = 95
```

Bu göstərici məlumatlarda çatışmayan dəyərlərin olub-olmadığını anlamağa kömək edir.

---

## `mean`

**mean** — ədədlərin orta qiymətidir.

Bütün qiymətlər toplanır və onların sayına bölünür.

Formula:

```text
mean = dəyərlərin cəmi / dəyərlərin sayı
```

Məsələn:

```text
10, 20, 30
```

```text
mean = (10 + 20 + 30) / 3 = 20
```

Orta qiymət məlumatların ümumi səviyyəsini göstərmək üçün istifadə olunur.

---

## `std`

**std** — standart kənarlaşmadır.

Dəyərlərin orta qiymətdən nə qədər uzaqda yerləşdiyini göstərir.

- `std` kiçikdirsə — dəyərlər bir-birinə və orta qiymətə yaxındır.
- `std` böyükdürsə — dəyərlər daha çox səpələnib.

Məsələn:

```text
10, 11, 12
```

Bu qiymətlər bir-birinə yaxın olduğu üçün standart kənarlaşma kiçik olacaq.

Amma:

```text
1, 50, 100
```

qiymətləri bir-birindən çox fərqləndiyi üçün standart kənarlaşma böyük olacaq.

---

## `min`

**min** — məlumatlardakı minimum, yəni ən kiçik qiyməti göstərir.

Məsələn:

```text
10, 20, 30, 5
```

Nəticə:

```text
min = 5
```

---

## `25%`

**25%** — birinci kvartildir və `Q1` kimi də işarə olunur.

Məlumatların təxminən `25%`-i bu qiymətdən kiçik və ya ona bərabər olur.

Məsələn:

```text
25% = 20
```

Bu o deməkdir ki, məlumatların təxminən dörddə biri `20` və ya daha aşağı qiymətlərdən ibarətdir.

---

## `50%`

**50%** — median və ya ikinci kvartildir (`Q2`).

Məlumatları kiçikdən böyüyə düzdükdə ortada yerləşən qiyməti göstərir.

Məlumatların:

- `50%`-i mediandan aşağı,
- `50%`-i mediandan yuxarı olur.

Məsələn:

```text
10, 20, 30, 40, 50
```

Burada:

```text
median = 30
```

Median çox böyük və ya çox kiçik ekstremal qiymətlər olduqda orta qiymətdən daha faydalı ola bilər.

---

## `75%`

**75%** — üçüncü kvartildir və `Q3` kimi işarə olunur.

Məlumatların təxminən `75%`-i bu qiymətdən kiçik və ya ona bərabər olur.

Başqa sözlə, məlumatların yalnız təxminən `25%`-i bu qiymətdən böyük olur.

Məsələn:

```text
75% = 80
```

Bu o deməkdir ki, məlumatların təxminən `75%`-i `80` və ya daha aşağıdır.

---

## `max`

**max** — məlumatlardakı maksimum, yəni ən böyük qiyməti göstərir.

Məsələn:

```text
10, 20, 30, 50
```

Nəticə:

```text
max = 50
```

---

## Qısa nümunə

Əgər qiymətlər belədirsə:

```text
10, 20, 30, 40, 50
```

`describe()` nəticəsi bizə bu məlumatları verə bilər:

| Göstərici | Mənası |
|---|---|
| `count` | Dəyərlərin sayı |
| `mean` | Orta qiymət |
| `std` | Standart kənarlaşma |
| `min` | Ən kiçik qiymət |
| `25%` | Birinci kvartil (`Q1`) |
| `50%` | Median (`Q2`) |
| `75%` | Üçüncü kvartil (`Q3`) |
| `max` | Ən böyük qiymət |
